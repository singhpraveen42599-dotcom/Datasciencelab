Q7.1. Write a function my_abs(val) that returns the absolute value of val without using Python's built-in abs().
Test it with a positive number, a negative number, and 0.

```python
def my_abs(val):
    if val < 0:
        return -val
    else:
        return val


# Test the function
print(my_abs(5))
print(my_abs(-5))
print(my_abs(0))
```
